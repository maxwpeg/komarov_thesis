```python
from reportlab.lib.pagesizes import A3, A4, landscape
from reportlab.lib.units import mm



print(mm)
```

```python
from reportlab.pdfgen import canvas

c = canvas.Canvas("test.pdf", A3)
c.
c.showPage()
c.save()
```

```python
# create a new PDF with several pages of different sizes and orientations
pdf = canvas.Canvas("multi_pages.pdf", pagesize=A4)

pages = [
    ("A4 portrait", A4),
    ("A4 landscape", landscape(A4)),
    ("A3 portrait", A3),
    ("A3 landscape", landscape(A3)),
]

for title, (w, h) in pages:
    pdf.setPageSize((w, h))
    w_mm = w / mm
    h_mm = h / mm

    pdf.setFont("Helvetica-Bold", 16)
    pdf.drawString(20 * mm, h - 20 * mm, title)

    pdf.setFont("Helvetica", 10)
    pdf.drawString(20 * mm, h - 30 * mm, f"Size: {w:.2f} x {h:.2f} points")
    pdf.drawString(20 * mm, h - 36 * mm, f"Size: {w_mm:.1f} x {h_mm:.1f} mm")
    pdf.drawString(20 * mm, h - 42 * mm, f"Orientation: {'landscape' if w > h else 'portrait'}")

    pdf.showPage()

pdf.save()
```
